Reviewer notebook for the predeclared audit sample drawn by `src/sample_audit.py --seed 297`:
300 questions per dataset (150 labelled *yes*, 150 labelled *no*, 50 from each split), shuffled.
The review is blind: you see only the image and the object name, never the dataset label or a model answer.

**Guidelines**
- **Present**: any clearly recognisable instance, including small, partial, or depicted (poster, screen, mirror, toy; tick *depicted*).
- **Absent**: confident it is not there after inspecting the whole image (use the zoom).
- **Uncertain**: too small, blurry or ambiguous to decide, or the category boundary is unclear (bench vs chair).

Every click saves `data/audit/labels_<dataset>_<reviewer>.csv`; close at any time and rerun the cell to resume.

One reviewer per item. To share the work, split it by list (for example one person does COCO, another GQA), not within a list: the tool always resumes at the first item *you* have not labelled. # Stage B: review of model errors (ordinary prompt)

Lists drawn by `src/sample_stage_b.py --seed 297` (false positives capped at 100 per dataset). Every item here is a model error, so you already know the dataset label; still judge only what is in the image.

**False positives** (`fp`: label *no*, model said *yes*). Verdict as in Stage A, plus one tag:
- *label error*: the object is actually there (filled in automatically when you click Present without a tag);
- *co-occurrence prior*: absent, but contextually expected (fork at a dinner table);
- *similar object*: something that looks like the probe (van taken for a bus);
- *depicted or reflected*: only in a picture, screen, mirror, or as a toy (tick *depicted* too);
- *ambiguous category*: unclear boundary (bench vs chair, cup vs mug);
- *other / no visible cause*.

**False negatives** (`fn`: label *yes*, model said *no*). Tags: *label error* (object not actually there; filled in automatically for Absent), *small*, *occluded or truncated*, *unusual appearance*, *other*.

Saved to `data/audit/labels_{fp,fn}_<dataset>_<reviewer>.csv`.
